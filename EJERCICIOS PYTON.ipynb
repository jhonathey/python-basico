```python
#30 Crea una función que determine si dos palabras son anagramas, es decir, si están formadas por las mismas letras 
# pero en diferente orden.

def son_anagramas(palabra1, palabra2):
  limpia1 = palabra1.lower().replace(" ", "")
  limpia2 = palabra2.lower().replace(" ", "")

  if len(limpia1) != len(limpia2):
    return False

  letras1 = []
  letras2 = []

  for letra in limpia1:
    letras1.append(letra)

  for letra in limpia2:
    letras2.append(letra)

  letras1.sort()
  letras2.sort()

  return letras1 == letras2

print(son_anagramas("Roma", "Amor"))
```

```python
#34 Crea la clase Arbol , define un árbol genérico con un tronco y ramas como atributos. Los métodos disponibles son:
# crecer_tronco , nueva_rama , crecer_ramas , quitar_rama e info_arbol . El objetivo es implementar estos métodos para
# manipular la estructura del árbol.
class Arbol:

  def __init__(self, tronco=1, ramas=None):
    self.tronco = tronco
    if ramas is None:
      self.ramas = []
    else:
      self.ramas = list(ramas)

  def crecer_tronco(self, incremento=1):
    self.tronco = self.tronco + incremento

  def nueva_rama(self, longitud=1):
    self.ramas.append(longitud)

  def crecer_ramas(self, incremento=1):
    for i in range(len(self.ramas)):
      self.ramas[i] = self.ramas[i] + incremento

  def quitar_rama(self, indice):
    if 0 <= indice < len(self.ramas):
      self.ramas.pop(indice)
    else:
      print(f"No existe una rama en el índice {indice}.")

  def info_arbol(self):
    return {
        "longitud_tronco": self.tronco,
        "cantidad_ramas": len(self.ramas),
        "longitudes_ramas": self.ramas,
    }



mi_arbol = Arbol() 
mi_arbol.crecer_tronco() 
mi_arbol.nueva_rama()  
mi_arbol.crecer_ramas() 
mi_arbol.nueva_rama(3) 
mi_arbol.nueva_rama(2)
mi_arbol.quitar_rama(1)  

resultado = mi_arbol.info_arbol()

print(resultado)
```

```python
#35Crea la clase UsuarioBanco ,representa a un usuario de un banco con su nombre, saldo y si tiene o no cuenta
# corriente. Proporciona métodos para realizar operaciones como retirar dinero, transferir dinero desde otro usuario
#  yagregar dinero al saldo.
# Código a seguir:
# 1. Inicializar un usuario con su nombre, saldo y si tiene o no cuenta corriente mediante True y False .
# 2. Implementar el método retirar_dinero para retirar dinero del saldo del usuario. Lanzará un error en caso de no
# poder hacerse.
# 3. Implementar el método transferir_dinero para realizar una transferencia desde otro usuario al usuario actual.
# Lanzará un error en caso de no poder hacerse.
# 4. Implementar el método agregar_dinero para agregar dinero al saldo del usuario.resultado
# 1. Crear dos usuarios: "Alicia" con saldo inicial de 100 y "Bob" con saldo inicial de 50, ambos con cuenta
#  corriente.
# PROYECTO LÓGICA: Katas de Python 3
# 2. Agregar 20 unidades de saldo de "Bob".
# 3. Hacer una transferencia de 80 unidades desde "Bob" a "Alicia".
# 4. Retirar 50 unidades de saldo a "Alicia".

class UsuarioBanco:

  def __init__(self, nombre, saldo, cuenta_corriente):
    self.nombre = nombre
    self.saldo = saldo
    self.cuenta_corriente = cuenta_corriente

  def retirar_dinero(self, cantidad):
    if cantidad <= 0:
      raise ValueError("La cantidad a retirar debe ser mayor que 0.")
    if cantidad > self.saldo:
      raise ValueError(
          f"{self.nombre} no tiene saldo suficiente para retirar {cantidad}."
      )
    self.saldo = self.saldo - cantidad

  def transferir_dinero(self, otro_usuario, cantidad):
    if cantidad <= 0:
      raise ValueError("La cantidad a transferir debe ser mayor que 0.")
    if not self.cuenta_corriente or not otro_usuario.cuenta_corriente:
      raise ValueError(
          "Ambos usuarios deben tener cuenta corriente para transferir."
      )
    if cantidad > otro_usuario.saldo:
      raise ValueError(
          f"{otro_usuario.nombre} no tiene saldo suficiente para transferir"
          f" {cantidad} a {self.nombre}."
      )
    otro_usuario.saldo = otro_usuario.saldo - cantidad
    self.saldo = self.saldo + cantidad

  def agregar_dinero(self, cantidad):
    if cantidad <= 0:
      raise ValueError("La cantidad a agregar debe ser mayor que 0.")
    self.saldo = self.saldo + cantidad



martha = UsuarioBanco("martha", 100, True)
jhonathey = UsuarioBanco("jhonathey", 50, True)


jhonathey.agregar_dinero(20)
print(f"Saldo de {jhonathey.nombre} tras agregar 20: {jhonathey.saldo}")


try:
  martha.transferir_dinero(jhonathey, 80)
except ValueError as error:
  print(f"Error en transferencia: {error}")


martha.retirar_dinero(50)
print(f"Saldo de {martha.nombre} tras retirar 50: {martha.saldo}")


```

```python
#37 Crea una función llamada procesar_texto que procesa un texto según la opción especificada: reemplazar_palabras
#, procesar_texto . contar_palabras , eliminar_palabra . Estas opciones son otras funciones que tenemos que definir
#primero y llamar dentro de la función Código a seguir: Crear una función contar_palabras para contar el número
#de veces que aparece cada palabra en el texto. Tiene que devolver un diccionario. Crear una función 
#reemplazar_palabras para remplazar una que devolver el texto con el remplazo de palabras. Crear una función 
#palabra_original del texto por una palabra_nueva . Tiene eliminar_palabra para eliminar una palabra del texto.
#Tiene que devolver el texto con la palabra eliminada. Crear la función procesar_texto que tome un texto, una 
#opción(entre "contar", "reemplazar", "eliminar") y un número de argumentos variable según la opción indicada.
#Caso de uso: Comprueba el funcionamiento completo de la función procesar_texto

import re


def contar_palabras(texto: str) -> dict[str, int]:
    
    palabras = texto.split()
    conteo = {}
    for palabra in palabras:
        conteo[palabra] = conteo.get(palabra, 0) + 1
    return conteo


def reemplazar_palabras(texto: str, palabra_original: str, palabra_nueva: str) -> str:
    return texto.replace(palabra_original, palabra_nueva)


def eliminar_palabra(texto: str, palabra_a_eliminar: str) -> str:
    patron = rf"\b{re.escape(palabra_a_eliminar)}\b"
    texto_sin_palabra = re.sub(patron, "", texto)
    
    return " ".join(texto_sin_palabra.split())


def procesar_texto(texto: str, opcion: str, *args):
    
    opcion_normalizada = opcion.strip().lower()

    if opcion_normalizada == "contar":
        return contar_palabras(texto)

    if opcion_normalizada == "reemplazar":
        if len(args) != 2:
            raise ValueError(
                "La opción 'reemplazar' requiere 2 argumentos: palabra_original y palabra_nueva."
            )
        palabra_original, palabra_nueva = args
        return reemplazar_palabras(texto, palabra_original, palabra_nueva)

    if opcion_normalizada == "eliminar":
        if len(args) != 1:
            raise ValueError(
                "La opción 'eliminar' requiere 1 argumento: palabra_a_eliminar."
            )
        (palabra_a_eliminar,) = args
        return eliminar_palabra(texto, palabra_a_eliminar)

    raise ValueError(
        f"Opción no válida: '{opcion}'. Usa 'contar', 'reemplazar' o 'eliminar'."
    )
frase = ("don victor mi mascota favoriota, en noviembre cumplio 2 años de edad, es guapo el don victor")


print("Contar:", procesar_texto(frase, "contar"))
print("Reemplazar:", procesar_texto(frase, "reemplazar", "victor", "gato"))
print("Eliminar:", procesar_texto(frase, "eliminar", "don"))
```

```python
# Genera un programa que nos diga si es de noche, de día o tarde según la hora proporcionada por el usuario.
def momento_del_dia(hora: int) -> str:
    """Devuelve 'día', 'tarde' o 'noche' según una hora válida (0 a 23)."""
    if not 0 <= hora <= 24:
        raise ValueError("La hora debe estar entre 0 y 24.")

    if 6 <= hora < 12:
        return "día"
    if 12 <= hora < 20:
        return "tarde"
    return "noche"


def main() -> None:
    entrada = input("Introduce la hora (0-24): ").strip()
    try:
        hora = int(entrada)
        resultado = momento_del_dia(hora)
        print(f"A las {hora}:00 es de {resultado}.")
    except ValueError as error:
        print(f"Error: {error}")


if __name__ == "__main__":
    main()
```

```python
#40 Escribe una función que tome dos parámetros: "triangulo" ) y figura (una cadena que puede ser "rectangulo" ,
#  "circulo" o datos (una tupla con los datos necesarios para calcular el área de la figura).

import math


def calcular_area(figura: str, datos: tuple) -> float:
    
    tipo = figura.strip().lower()

    if tipo == "triangulo":
        if len(datos) != 2:
            raise ValueError("El triángulo requiere 2 datos: (base, altura).")
        base, altura = datos
        if base <= 0 or altura <= 0:
            raise ValueError("La base y la altura deben ser mayores que 0.")
        return (base * altura) / 2

    if tipo == "rectangulo":
        if len(datos) != 2:
            raise ValueError("El rectángulo requiere 2 datos: (base, altura).")
        base, altura = datos
        if base <= 0 or altura <= 0:
            raise ValueError("La base y la altura deben ser mayores que 0.")
        return base * altura

    if tipo == "circulo":
        if len(datos) != 1:
            raise ValueError("El círculo requiere 1 dato: (radio,).")
        (radio,) = datos
        if radio <= 0:
            raise ValueError("El radio debe ser mayor que 0.")
        return math.pi * (radio**2)

    raise ValueError(
        f"Figura no válida: '{figura}'. Usa 'triangulo', 'rectangulo' o 'circulo'."
    )



print(calcular_area("triangulo", (6, 4)))   
print(calcular_area("rectangulo", (6, 4)))  
print(calcular_area("circulo", (5,)))       
```
